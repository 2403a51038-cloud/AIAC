Task Description#1
●	Zero-shot: Prompt AI to write a function that checks whether a given year is a leap year.
Expected Output#1
●	AI-generated function with no examples provided




```python
def is_leap(year):
  """
  Checks if a given year is a leap year.

  Args:
    year: The year to check (integer).

  Returns:
    True if the year is a leap year, False otherwise.
  """
  if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    return True
  else:
    return False

# Get input from the user
year_str = input("Enter a year: ")

try:
  year = int(year_str)
  if is_leap(year):
    print(f"{year} is a leap year.")
  else:
    print(f"{year} is not a leap year.")
except ValueError:
  print("Invalid input. Please enter a valid year (a number).")
```

Task Description#2
●	One-shot: Give one input-output example to guide AI in writing a function that converts centimeters to inches.
Expected Output#2
●	Function with correct conversion logic


```python
def cm_to_inches(cm):
  """
  Converts centimeters to inches.

  Args:
    cm: The length in centimeters (float).

  Returns:
    The length in inches (float).
  """
  return cm * 0.393701

# Get input from the user
cm_str = input("Enter length in centimeters: ")

try:
  cm = float(cm_str)
  inches = cm_to_inches(cm)
  print(f"{cm}cm = {inches:.2f} inches")
except ValueError:
  print("Invalid input. Please enter a valid number for centimeters.")
```

Task Description#3
●	Few-shot: Provide 2–3 examples to generate a function that formats full names as “Last, First”.
Expected Output#3
●	Well-structured function respecting the examples


```python
def format_name(full_name):
  """
  Formats a full name as 'Last, First'.

  Args:
    full_name: The full name (string) with first and last name separated by a space.

  Returns:
    The formatted name as 'Last, First' (string), or the original name if it doesn't contain a space.
  """
  name_parts = full_name.split()
  if len(name_parts) >= 2:
    return f"{name_parts[-1]}, {' '.join(name_parts[:-1])}"
  else:
    return full_name # Return original name if it doesn't have a space

# Get input from the user
full_name_str = input("Enter a full name (e.g., First Last): ")

formatted_name = format_name(full_name_str)
print(f"Formatted name: {formatted_name}")
```

Task Description#4
●	Compare zero-shot and few-shot prompts for writing a function that counts the number of vowels in a string.
Expected Output#4
●	Functional output and comparative reflection


```python
def count_vowels(text):
  """
  Counts the number of vowels in a string.

  Args:
    text: The input string.

  Returns:
    The number of vowels in the string.
  """
  vowels = "aeiouAEIOU"
  vowel_count = 0
  for char in text:
    if char in vowels:
      vowel_count += 1
  return vowel_count

# Get input from the user
input_string = input("Enter a string: ")

vowel_count = count_vowels(input_string)
print(f"Number of vowels in a given string : {vowel_count}")
```

Task Description#5
●	Use few-shot prompting to generate a function that reads a .txt file and returns the number of lines.
Expected Output#5
●	Working file-processing function with AI-guided logic


```python
def count_lines_in_file(filename):
  """
  Reads a file and returns the number of lines.

  Args:
    filename: The name of the text file (assuming it's in the Colab environment).

  Returns:
    The number of lines in the file, or an error message if the file cannot be read.
  """
  try:
    with open(filename, 'r') as f:
      lines = f.readlines()
      return len(lines)
  except FileNotFoundError:
    return f"Error: File not found. Please ensure '{filename}' is uploaded to your Colab environment."
  except Exception as e:
    return f"An error occurred: {e}"

# Specify the filename
file_name = "lab 4.txt"

line_count = count_lines_in_file(file_name)

print(f"Number of lines in '{file_name}': {line_count}")
```
